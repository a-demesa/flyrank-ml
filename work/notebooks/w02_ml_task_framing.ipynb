## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a **scoring** task. The goal is to assign each webpage a score that represents how strongly it should be prioritized for a content refresh. Instead of making a simple yes-or-no decision, scoring allows pages to be ranked from highest to lowest priority so content teams can focus on the most valuable pages first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is a content refresh priority score. Since there is no direct label showing whether a page should be refreshed, a proxy can be created from observed signals such as traffic trend, click-through rate (CTR), and other available features in the dataset. These observed values help estimate which pages are more likely to benefit from a content update.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

A suitable success metric is **Precision@50** because the goal is to rank the pages most likely to benefit from a content refresh. A higher Precision@50 means more of the top-ranked pages are good refresh candidates, helping content teams spend their time on the right pages.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule here

A fixed rule might only check one or two conditions, such as whether a page is old or has many impressions. However, deciding which pages should be refreshed depends on several factors working together, including impressions, click-through rate (CTR), average search position, content age, and traffic trends. These relationships are difficult to capture with a single if-statement. Machine learning can learn patterns from multiple features at the same time and produce a more reliable ranking of which pages should be prioritized for content refresh.
