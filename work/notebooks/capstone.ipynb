## Dataset

This project uses the FlyRank ML Internship dataset based on anonymized website search performance records.

One row represents the daily search performance of one content page for one client on a specific report date.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## Assumptions

This project assumes that observed search performance metrics provide useful signals for identifying webpages that may benefit from content review.

## Features

- GSC Impressions
- GSC Clicks
- Average Search Position

## Target

The target represents whether a webpage receives search clicks and serves as a simplified proxy for content performance.

## Baseline

A rule-based ranking approach developed during Week 4 was used as the baseline.

## Model

A Decision Tree classifier was selected because it is easy to interpret and appropriate for a simple baseline comparison.

## Validation

A grouped validation split based on client identifiers was used to reduce the chance of information leakage between training and testing.

## Leakage Check

Feature selection was reviewed to avoid using future information or variables that directly define the prediction target.

## 5. Limitations

*What this work cannot claim.*

This project is based on observed search performance data and cannot determine why performance changes occur.

The model does not consider:

- content quality
- competitor activity
- search engine algorithm updates
- business priorities

The results should therefore be interpreted as decision-support rather than causal evidence.
