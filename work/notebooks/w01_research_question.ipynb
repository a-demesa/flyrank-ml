```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))
print("Pages with downward trend:", (df["trend_direction"] == "down").sum())
print("Average CTR:", round(df["ctr"].mean(), 3))
```

Out of these, **16,262 pages** have a downward trend, indicating that many pages could potentially benefit from a content refresh.

The average click-through rate (CTR) is **0.511**, showing that page performance varies and that multiple signals can be used to prioritize which pages should be reviewed first.
