## 2. My model under an honest split (before/after)

*Re-run your Week-5 model under a grouped or time-aware split. Show both numbers.*

I re-ran my Week 5 Decision Tree model using a grouped split based on `client_hash_id` instead of only a random train-test split. Grouping by client reduces the chance that similar observations appear in both the training and testing datasets, giving a more realistic estimate of model performance. This validation approach better reflects how the model would perform on unseen clients and supports more trustworthy evaluation results.

```python
from sklearn.model_selection import GroupShuffleSplit
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
```

```python
from datasets import load_dataset

dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance"
)

train = dataset["train"]

sample_df = train.select(range(10000)).to_pandas()

sample_df.head()
```

```python
features = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]

df = sample_df.dropna(subset=features).copy()

df["target"] = (df["gsc_clicks"] > 0).astype(int)
```

```python
splitter = GroupShuffleSplit(
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        df[features],
        df["target"],
        groups=df["client_hash_id"]
    )
)
```

```python
X_train = df.iloc[train_idx][features]
X_test = df.iloc[test_idx][features]

y_train = df.iloc[train_idx]["target"]
y_test = df.iloc[test_idx]["target"]
```

```python
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)

pred = model.predict(X_test)

group_accuracy = accuracy_score(
    y_test,
    pred
)

print(group_accuracy)
```

```python
comparison = pd.DataFrame({
    "Validation Method": [
        "Random Split (Week 5)",
        "Grouped by Client"
    ],
    "Accuracy": [
        1.00,
        group_accuracy
    ]
})

comparison
```
