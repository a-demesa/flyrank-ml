## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

This playbook ranks webpages using observed search performance signals to help prioritize content review.

| Priority | Reason Code | Recommended Action |
|----------|-------------|--------------------|
| High | REFRESH_HIGH | Review and refresh page content |
| Medium | REFRESH_MEDIUM | Review content quality and keyword coverage |
| Low | MONITOR | Continue monitoring performance |
| Lowest | NO_ACTION | No immediate action required |

```python
import pandas as pd

queue = pd.DataFrame({
    "content_hash_id": sample_df["content_hash_id"].head(10),
    "score": sample_df["gsc_impressions"].head(10)
})

queue["reason_code"] = "REFRESH_HIGH"
queue["recommended_action"] = "Review and refresh content"

queue
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## Intended use

This playbook is intended to help content teams prioritize webpages that may benefit from review based on observed search performance signals.

## Limits

These recommendations are decision-support only. They do not account for content quality, business priorities, competitor activity, or search engine algorithm updates. Human review is required before any action is taken.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human review

Before acting on any recommendation, a reviewer should confirm:

- The page is still relevant.
- The content is accurate.
- The recommendation aligns with business priorities.
- User intent has not changed.

## No-go list

This system should never automatically:

- Publish content
- Delete webpages
- Rewrite content without review
- Replace human editorial judgment
- Make business decisions

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring and retraining

The model should be reviewed whenever:

- Prediction quality decreases.
- Search behavior changes noticeably.
- New training data becomes available.
- Business goals or SEO strategy change.

Regular evaluation helps keep recommendations relevant over time.
